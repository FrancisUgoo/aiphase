```python
from openai import OpenAI
openai_client = OpenAI()
print("OpenAI client created")
```

```python
response = openai_client.responses.create(model="gpt-4o-mini", input="Write a short bedtime story about a unicorn.")

print(response.output_text)
```

```python
response2 = openai_client.responses.create(
    model="gpt-4o-mini", input="Tell me story about a man named Francis with good works. not more than 30 words"
)
print(response2.output_text)
```

```python
response3 = openai_client.responses.create(
    model="gpt-4o-mini",
    input=[
        {"role": "developer", "content": "Talk like a pirate."},
        {"role": "system", "content": "Make sure to use emojis."},
        {"role": "user", "content": "Are semicolons optional in JavaScript?"},
    ],
)

print(response3.output_text)
```

```python
# print(response3.output_text) is a shortcut/equal to print(response3.output[0].content[0].text) 

print(response3.output[0].content[0].text)
```

```python
print(response3.usage)
#see the input_tokens and output_tokens values for pricing effect for this choosen model. model="gpt-4o-mini"
```

```python
stream_response = openai_client.responses.create(
    model="gpt-4o-mini",
    input=[
        {"role": "developer", "content": "Talk like a pirate. Response must not be more than 30 words."},
        {"role": "system", "content": "Make sure to use emojis."},
        {"role": "user", "content": "Write a short bedtime story about a unicorn."},
    ],
    stream=True
)

for chunk in stream_response:
    if hasattr(chunk, 'delta'):
        print(chunk.delta, end='', flush=True) #flush not necessarily needed. end="\n" is default that is why we need it to print continously, so end with nothing ''


# print(stream_response.output_text)
```

```python
stream_response2 = openai_client.responses.create(
    model="gpt-4o-mini",
    input=[
        {"role": "developer", "content": "Talk like a pirate. Response must not be more than 30 words."},
        {"role": "system", "content": "Make sure to use emojis."},
        {"role": "user", "content": "Write a short bedtime story about a unicorn."},
    ],
    stream=True
)

chunks_arr=[]

for chunk in stream_response2:
    chunks_arr.append(chunk)
    if hasattr(chunk, 'delta'):
        print(chunk.delta, end='', flush=True) 
```

```python
print(chunks_arr[-1].response.model)
print('----------------------------------------------')
chunks_arr[-1].response.output[0].content[0].text
print(chunks_arr[-1].response.output[0].content[0].text, end='', flush=True) 
print('\n ----------------------------------------------')
print(chunks_arr[-1].response.usage)
```
